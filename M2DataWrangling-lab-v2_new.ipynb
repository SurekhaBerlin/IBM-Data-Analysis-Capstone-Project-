### 3. Identifying and Removing Inconsistencies


<h5>3.1 Identify inconsistent or irrelevant entries in specific columns (e.g., Country).</h5>


```python
import pandas as pd

# Sample DataFrame
df = pd.DataFrame({'Country': ['Canada', 'US', 'Cnada', 'Germany', '123', '', 'U.S.A.']})

# List of valid countries
valid_countries = ['Canada', 'United States', 'Germany']

# Identify inconsistent/irrelevant entries
df['Invalid_Country'] = ~df['Country'].isin(valid_countries)
print(df[df['Invalid_Country']])
```

<h5>3.2 Standardize entries in columns like Country or EdLevel by mapping inconsistent values to a consistent format.</h5>


```python
import pandas as pd

# Sample DataFrame
df = pd.DataFrame({'Country': ['US','U.S.A.','Cnada','Canada','Germany'],
                   'EdLevel': ['BSc','Bachelor’s','MSc','Master’s Degree','PhD']})

# Mapping dictionaries
country_map = {
    'US': 'United States',
    'U.S.A.': 'United States',
    'Cnada': 'Canada'
}

edlevel_map = {
    'BSc': 'Bachelor',
    'Bachelor’s': 'Bachelor',
    'MSc': 'Master',
    'Master’s Degree': 'Master',
    'PhD': 'Doctorate'
}

# Apply mapping
df['Country'] = df['Country'].replace(country_map)
df['EdLevel'] = df['EdLevel'].replace(edlevel_map)

print(df)
```

### 4. Encoding Categorical Variables


<h5>4.1 Encode the Employment column using one-hot encoding.</h5>


```python
import pandas as pd

# Sample DataFrame
df = pd.DataFrame({'Employment': ['Full-time', 'Part-time', 'Self-employed', 'Unemployed', 'Full-time']})

# One-hot encoding
df_encoded = pd.get_dummies(df, columns=['Employment'])

print(df_encoded)
```

### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>


```python
import pandas as pd

# Sample DataFrame
df = pd.DataFrame({
    'Country': ['Canada', None, 'US', 'Germany', None],
    'EdLevel': ['BSc', 'MSc', None, 'PhD', 'Master’s'],
    'Employment': ['Full-time', None, 'Part-time', 'Self-employed', 'Full-time']
})

# Count missing values per column
missing_counts = df.isna().sum().sort_values(ascending=False)

print(missing_counts)
```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


```python
import pandas as pd
import numpy as np

# Sample DataFrame
df = pd.DataFrame({
    'ConvertedCompYearly': [50000, 60000, np.nan, 55000, 120000]
})

# Impute with mean
mean_value = df['ConvertedCompYearly'].mean()
df['ConvertedCompYearly_mean'] = df['ConvertedCompYearly'].fillna(mean_value)

# Impute with median
median_value = df['ConvertedCompYearly'].median()
df['ConvertedCompYearly_median'] = df['ConvertedCompYearly'].fillna(median_value)

print(df)
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
import pandas as pd
import numpy as np

# Sample DataFrame
df = pd.DataFrame({
    'RemoteWork': ['Fully remote', 'Partially remote', np.nan, 'Fully remote', 'Onsite']
})

# Find mode
mode_value = df['RemoteWork'].mode()[0]

# Impute missing values
df['RemoteWork'] = df['RemoteWork'].fillna(mode_value)

print(df)
```

### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


```python
!pip install scikit-learn
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Sample DataFrame
df = pd.DataFrame({
    'ConvertedCompYearly': [50000, 60000, 70000, 100000]
})

# Initialize scaler
scaler = MinMaxScaler()

# Apply Min-Max scaling
df['ConvertedCompYearly_scaled'] = scaler.fit_transform(df[['ConvertedCompYearly']])

print(df)

```

<h5>6.2 Log-transform the ConvertedCompYearly column to reduce skewness.</h5>


```python
import pandas as pd
import numpy as np

# Sample DataFrame
df = pd.DataFrame({
    'ConvertedCompYearly': [0, 50000, 100000, 500000]
})

# Apply natural log transformation
df['ConvertedCompYearly_log'] = np.log(df['ConvertedCompYearly'] + 1)

# Optional: log10
df['ConvertedCompYearly_log10'] = np.log10(df['ConvertedCompYearly'] + 1)

print(df)
```

### 7. Feature Engineering


<h5>7.1 Create a new column `ExperienceLevel` based on the `YearsCodePro` column:</h5>


```python
import pandas as pd

# Sample DataFrame
df = pd.DataFrame({'YearsCodePro': [0.5, 2, 4, 7, 12]})

# Define bins and labels
bins = [0, 1, 3, 5, 10, float('inf')]
labels = ['Beginner', 'Junior', 'Intermediate', 'Senior', 'Expert']

# Create ExperienceLevel column
df['ExperienceLevel'] = pd.cut(df['YearsCodePro'], bins=bins, labels=labels, right=True)

print(df)
```
