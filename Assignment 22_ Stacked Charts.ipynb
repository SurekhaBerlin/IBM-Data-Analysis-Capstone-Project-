### Task 1: Stacked Chart for Composition of Job Satisfaction Across Age Groups


##### 1. Stacked Chart of Median `JobSatPoints_6` and `JobSatPoints_7` for Different Age Groups


Visualize the composition of job satisfaction scores (`JobSatPoints_6` and `JobSatPoints_7`) across various age groups. This will help in understanding the breakdown of satisfaction levels across different demographics.



```python
import pandas as pd
import matplotlib.pyplot as plt

# Group data by Age and calculate median for JobSatPoints_6 and JobSatPoints_7
median_scores = df.groupby('Age')[['JobSatPoints_6', 'JobSatPoints_7']].median()

# Optional: Check the aggregated table
print(median_scores)
```

```python
# Plot stacked bar chart
median_scores.plot(kind='bar', stacked=True, figsize=(10,6), color=['#1f77b4', '#ff7f0e'])

plt.title('Median Job Satisfaction Scores Across Age Groups')
plt.xlabel('Age Group')
plt.ylabel('Median Job Satisfaction Points')
plt.legend(title='Job Satisfaction Points')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

##### Stacked Chart of `JobSatPoints_6` and `JobSatPoints_7` for Employment Status


Create a stacked chart to compare job satisfaction (`JobSatPoints_6` and `JobSatPoints_7`) across different employment statuses. This will show how satisfaction varies by employment type.


```python
import pandas as pd
import matplotlib.pyplot as plt

# Group by Employment and calculate median for JobSatPoints_6 and JobSatPoints_7
median_scores_status = df.groupby('Employment')[['JobSatPoints_6', 'JobSatPoints_7']].median()

# Select the top 10 employment types by total median satisfaction (JobSatPoints_6 + JobSatPoints_7)
median_scores_status['Total'] = median_scores_status['JobSatPoints_6'] + median_scores_status['JobSatPoints_7']
top10_status = median_scores_status.nlargest(10, 'Total').drop(columns='Total')

# Plot stacked bar chart
top10_status.plot(kind='bar', stacked=True, figsize=(12,6), color=['#2ca02c', '#d62728'])

plt.title('Median Job Satisfaction Scores by Top 10 Employment Types')
plt.xlabel('Employment Type')
plt.ylabel('Median Job Satisfaction Points')
plt.legend(title='Job Satisfaction Points')
plt.xticks(rotation=45, ha='right')  # Rotate labels for readability
plt.tight_layout()
plt.show()

```

### Task 2: Stacked Chart for Compensation and Job Satisfaction by Age Group


##### This stacked chart visualizes the composition of compensation (`ConvertedCompYearly`) and job satisfaction (`JobSatPoints_6`) specifically for respondents aged 30-35.


```python
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
age_group = df[(df['Age'] >= 30) & (df['Age'] <= 35)]
print(f"Number of respondents aged 30-35: {len(age_group)}")
```

```python
# Filter age groups labeled '30-35'
age_group = df[df['Age'] == '30-35']

# Now aggregate median values
median_age = age_group[['ConvertedCompYearly', 'JobSatPoints_6']].median().to_frame().T

# Optional: Convert compensation to thousands
median_age['ConvertedCompYearly'] = median_age['ConvertedCompYearly'] / 1000

# Plot as single stacked bar
median_age.plot(kind='bar', stacked=True, figsize=(6,6), color=['#1f77b4', '#ff7f0e'])
plt.title('Median Compensation (in 1000 USD) and Job Satisfaction (Age 30-35)')
plt.xlabel('Age Group 30-35')
plt.ylabel('Median Values')
plt.legend(title='Metric', labels=['Compensation (1000 USD)', 'Job Satisfaction Points'])
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

##### Stacked Chart of Median Compensation and Job Satisfaction Across Age Group


Compare the median compensation and job satisfaction metrics across different age groups. This helps visualize how compensation and satisfaction levels differ by age.


```python
##Write your code hereimport pandas as pd
import matplotlib.pyplot as plt

# Step 1: Convert Age to numeric if possible; keep non-numeric as NaN
df['Age_numeric'] = pd.to_numeric(df['Age'], errors='coerce')

# Step 2: Filter out rows where Age or metrics are missing
df_filtered = df[['Age_numeric', 'ConvertedCompYearly', 'JobSatPoints_6']].dropna()

# Step 3: Group by Age and calculate median values
median_by_age = df_filtered.groupby('Age_numeric')[['ConvertedCompYearly', 'JobSatPoints_6']].median()

# Optional: Convert compensation to thousands for readability
median_by_age['ConvertedCompYearly'] = median_by_age['ConvertedCompYearly'] / 1000

# Step 4: Plot stacked bar chart
median_by_age.plot(kind='bar', stacked=True, figsize=(12,6), color=['#1f77b4', '#ff7f0e'])

plt.title('Median Compensation (in 1000 USD) and Job Satisfaction by Age Group')
plt.xlabel('Age')
plt.ylabel('Median Values')
plt.legend(title='Metric', labels=['Compensation (1000 USD)', 'Job Satisfaction Points'])
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### Task 3: Comparing Data Using Stacked Charts


##### 1. Stacked Chart of Preferred Databases by Age Group




Visualize the top databases that respondents from different age groups wish to learn. Create a stacked chart to show the proportion of each database in each age group.


```python
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Convert Age to numeric (if survey uses numeric ages)
df['Age_numeric'] = pd.to_numeric(df['Age'], errors='coerce')

# Step 2: Filter rows where Age or database preference is missing
df_filtered = df[['Age_numeric', 'DatabaseWantToWorkWith']].dropna()

# Step 3: Count top N databases per age group
# Here we take top 5 databases overall for clarity
top_databases = df_filtered['DatabaseWantToWorkWith'].value_counts().nlargest(5).index
df_top = df_filtered[df_filtered['DatabaseWantToWorkWith'].isin(top_databases)]

# Step 4: Create a pivot table: rows=Age, columns=Database, values=counts
pivot_table = df_top.pivot_table(
    index='Age_numeric',
    columns='DatabaseWantToWorkWith',
    aggfunc='size',
    fill_value=0
)

# Step 5: Plot stacked bar chart
pivot_table.plot(kind='bar', stacked=True, figsize=(12,6), colormap='tab20')

plt.title('Top 5 Preferred Databases by Age Group')
plt.xlabel('Age')
plt.ylabel('Number of Respondents')
plt.legend(title='Database', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

##### 2. Stacked Chart of Employment Type by Job Satisfaction


Analyze the distribution of employment types within each job satisfaction level using a stacked chart. This will provide insights into how employment types are distributed across various satisfaction ratings.


```python
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Filter relevant columns and drop missing values
df_filtered = df[['Employment', 'JobSatPoints_6']].dropna()

# Step 2: Convert JobSatPoints_6 to integer if needed
df_filtered['JobSatPoints_6'] = pd.to_numeric(df_filtered['JobSatPoints_6'], errors='coerce')
df_filtered = df_filtered.dropna()

# Step 3: Create a pivot table: rows = Job Satisfaction, columns = Employment, values = counts
pivot_table = df_filtered.pivot_table(
    index='JobSatPoints_6',
    columns='Employment',
    aggfunc='size',
    fill_value=0
)

# Step 4: Plot stacked bar chart
pivot_table.plot(kind='bar', stacked=True, figsize=(12,6), colormap='tab20')

plt.title('Employment Type Distribution by Job Satisfaction')
plt.xlabel('Job Satisfaction Points')
plt.ylabel('Number of Respondents')
plt.legend(title='Employment Type', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### Task 4: Exploring Technology Preferences Using Stacked Charts


##### 1. Stacked Chart for Preferred Programming Languages by Age Group


Analyze how programming language preferences (`LanguageAdmired`) vary across age groups.


```python
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Convert Age to numeric (if survey uses numeric ages)
df['Age_numeric'] = pd.to_numeric(df['Age'], errors='coerce')

# Step 2: Filter rows with valid Age and LanguageAdmired
df_filtered = df[['Age_numeric', 'LanguageAdmired']].dropna()

# Step 3: Select top N languages overall (e.g., top 5) for clarity
top_languages = df_filtered['LanguageAdmired'].value_counts().nlargest(5).index
df_top = df_filtered[df_filtered['LanguageAdmired'].isin(top_languages)]

# Step 4: Create a pivot table: rows=Age, columns=Language, values=counts
pivot_table = df_top.pivot_table(
    index='Age_numeric',
    columns='LanguageAdmired',
    aggfunc='size',
    fill_value=0
)

# Step 5: Plot stacked bar chart
pivot_table.plot(kind='bar', stacked=True, figsize=(12,6), colormap='tab20')

plt.title('Top 5 Admired Programming Languages by Age Group')
plt.xlabel('Age')
plt.ylabel('Number of Respondents')
plt.legend(title='Programming Language', bbox_to_anchor=(1.05,1), loc='upper left')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

##### 2. Stacked Chart for Technology Adoption by Employment Type


Explore how admired platforms (`PlatformAdmired`) differ across employment types (e.g., full-time, freelance)


```python
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Filter relevant columns and drop missing values
df_filtered = df[['Employment', 'PlatformAdmired']].dropna()

# Step 2: Select top N admired platforms overall for readability (e.g., top 5)
top_platforms = df_filtered['PlatformAdmired'].value_counts().nlargest(5).index
df_top = df_filtered[df_filtered['PlatformAdmired'].isin(top_platforms)]

# Step 3: Create a pivot table: rows=Employment, columns=Platform, values=counts
pivot_table = df_top.pivot_table(
    index='Employment',
    columns='PlatformAdmired',
    aggfunc='size',
    fill_value=0
)

# Step 4: Plot stacked bar chart
pivot_table.plot(kind='bar', stacked=True, figsize=(12,6), colormap='tab20')

plt.title('Top 5 Admired Platforms by Employment Type')
plt.xlabel('Employment Type')
plt.ylabel('Number of Respondents')
plt.legend(title='Platform', bbox_to_anchor=(1.05,1), loc='upper left')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
