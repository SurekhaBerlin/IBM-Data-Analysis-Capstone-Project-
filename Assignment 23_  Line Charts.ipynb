### Task 1: Trends in Compensation Over Age Groups


##### 1. Line Chart of Median `ConvertedCompYearly` by Age Group


- Track how the median yearly compensation (ConvertedCompYearly) changes across different age groups.

- Use a line chart to visualize these trends.


```python
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Convert Age to numeric (ignore non-numeric ages)
df['Age_numeric'] = pd.to_numeric(df['Age'], errors='coerce')

# Step 2: Filter rows with valid Age and ConvertedCompYearly
df_filtered = df[['Age_numeric', 'ConvertedCompYearly']].dropna()

# Step 3: Group by Age and calculate median compensation
median_comp = df_filtered.groupby('Age_numeric')['ConvertedCompYearly'].median()

# Optional: Convert compensation to thousands for readability
median_comp = median_comp / 1000

# Step 4: Plot line chart
plt.figure(figsize=(12,6))
plt.plot(median_comp.index, median_comp.values, marker='o', color='#1f77b4')
plt.title('Median Yearly Compensation by Age Group')
plt.xlabel('Age')
plt.ylabel('Median Compensation (in 1000 USD)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

```

### Task 2: Trends in Job Satisfaction by Experience Level



##### 1. Line Chart of Job Satisfaction (`JobSatPoints_6`) by Experience Level



- Use a column that approximates experience level to analyze how job satisfaction changes with experience.

- If needed, substitute an available experience-related column for `Experience`.


```python
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Convert Age to numeric (used as experience proxy)
df['Age_numeric'] = pd.to_numeric(df['Age'], errors='coerce')

# Step 2: Filter rows with valid Age and JobSatPoints_6
df_filtered = df[['Age_numeric', 'JobSatPoints_6']].dropna()

# Step 3: Group by Age and calculate median Job Satisfaction
median_jobsat = df_filtered.groupby('Age_numeric')['JobSatPoints_6'].median()

# Step 4: Plot line chart
plt.figure(figsize=(12,6))
plt.plot(median_jobsat.index, median_jobsat.values, marker='o', color='#2ca02c')
plt.title('Median Job Satisfaction by Experience Level (Age Proxy)')
plt.xlabel('Age (as Experience Proxy)')
plt.ylabel('Median Job Satisfaction Points')
plt.grid(True, linestyle='--', alpha=0.5)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
