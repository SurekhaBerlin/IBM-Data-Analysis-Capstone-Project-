```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Optional: set seaborn style for prettier plots
sns.set(style="whitegrid")
%matplotlib inline
```

```python
# Load your Excel file
file_path = "/content/github_job_postings.xlsx"  # adjust path if needed
df = pd.read_excel(file_path)

# Preview the data
df.head()
```

```python
df_sorted = df.sort_values(by='Job Postings', ascending=False)
df_sorted
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='Job Postings', y='Language', data=df_sorted,  palette="Blues")
plt.title("Job Postings by Programming Language", fontsize=16)
plt.xlabel("Number of Job Postings", fontsize=12)
plt.ylabel("Programming Language", fontsize=12)
plt.show()
```
