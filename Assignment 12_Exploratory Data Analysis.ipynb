#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
# Keep only relevant columns
lang_df = df[['Country', 'LanguageHaveWorkedWith']].dropna()

# Split languages into individual rows
lang_df = lang_df.assign(
    Language=lang_df['LanguageHaveWorkedWith'].str.split(';')
).explode('Language')
```

```python
# Prepare language data
lang_df = df[['Country', 'LanguageHaveWorkedWith']].dropna()

lang_df = lang_df.assign(
    Language=lang_df['LanguageHaveWorkedWith'].str.split(';')
).explode('Language')

# Define regions
regions = {
    'North America': ['United States of America', 'Canada'],
    'Europe': ['Germany', 'United Kingdom', 'France'],
    'Asia': ['India', 'China', 'Japan']
}

lang_df['Region'] = lang_df['Country'].map(
    lambda x: next((r for r, c in regions.items() if x in c), None)
)

lang_df = lang_df.dropna(subset=['Region'])

# Count languages by region
language_counts = (
    lang_df
    .groupby(['Region', 'Language'])
    .size()
    .reset_index(name='Count')
)
```

```python
language_pivot = (
    language_counts
    .pivot(index='Language', columns='Region', values='Count')
    .fillna(0)
)
```

```python
top_10_languages = (
    language_counts
    .groupby('Language')['Count']
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .index
)

language_pivot = language_pivot.loc[top_10_languages]
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
plt.imshow(language_pivot)
plt.xticks(
    ticks=range(len(language_pivot.columns)),
    labels=language_pivot.columns,
    rotation=45
)
plt.yticks(
    ticks=range(len(language_pivot.index)),
    labels=language_pivot.index
)
plt.colorbar(label='Number of Developers')
plt.title("Top 10 Programming Languages by Region")
plt.tight_layout()
plt.show()
```

#### Step 8: Correlation Between Experience and Satisfaction


Examine how years of experience (`YearsCodePro`) correlate with job satisfaction (`JobSatPoints_1`). Use a scatter plot to visualize this relationship.


```python
import matplotlib.pyplot as plt
import pandas as pd

# Convert to numeric (safe conversion)
df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')
df['JobSatPoints_1'] = pd.to_numeric(df['JobSatPoints_1'], errors='coerce')

# Keep only valid rows
corr_df = df[['YearsCodePro', 'JobSatPoints_1']].dropna()
```

```python
plt.figure()
plt.scatter(
    corr_df['YearsCodePro'],
    corr_df['JobSatPoints_1'],
    alpha=0.5
)
plt.xlabel("Years of Professional Coding Experience")
plt.ylabel("Job Satisfaction Score")
plt.title("Relationship Between Experience and Job Satisfaction")
plt.show()
```
