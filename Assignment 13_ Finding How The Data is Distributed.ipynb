### Step 1: Import Libraries and Load Data


- Import the `pandas`, `matplotlib.pyplot`, and `seaborn` libraries.

```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Stack Overflow survey dataset
data_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv'
df = pd.read_csv(data_url)

# Display the first few rows of the dataset
df.head()
```

### Step 3: Handle Missing Data


- Identify missing values in the dataset.

- Impute or remove missing values as necessary to ensure data completeness.



```python
# Count missing values per column
missing_values = df.isnull().sum()
print("Missing values per column:")
print(missing_values)

# Optional: percentage of missing values
missing_percentage = (df.isnull().sum() / len(df)) * 100
missing_summary = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage (%)": missing_percentage
})
print("\nPercentage of missing values per column:")
print(missing_summary)
```

```python
categorical_cols = df.select_dtypes(include='object').columns

for col in categorical_cols:
    if df[col].isnull().sum() > 0:
        df[col].fillna(df[col].mode()[0], inplace=True)
```

```python
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns

for col in numeric_cols:
    if df[col].isnull().sum() > 0:
        df[col].fillna(df[col].median(), inplace=True)
```

```python
print("Missing values after imputation:")
print(df.isnull().sum())
```

### Step 4: Analyze Key Columns


- Examine key columns such as `Employment`, `JobSat` (Job Satisfaction), and `YearsCodePro` (Professional Coding Experience).

- **Instruction**: Calculate the value counts for each column to understand the distribution of responses.



```python
# Value counts for Employment
employment_counts = df['Employment'].value_counts(dropna=False)
print("Distribution of Employment:")
print(employment_counts)
```

```python
# Value counts for JobSat
jobsat_counts = df['JobSat'].value_counts(dropna=False)
print("\nDistribution of Job Satisfaction (JobSat):")
print(jobsat_counts)
```

### Step 6: Programming Languages Analysis


- Compare the frequency of programming languages in `LanguageHaveWorkedWith` and `LanguageWantToWorkWith`.
  
- Visualize the overlap or differences using a Venn diagram or a grouped bar chart.


```python
import pandas as pd

# Extract all languages worked with
have_worked = df['LanguageHaveWorkedWith'].dropna().str.split(';')
have_worked = [lang for sublist in have_worked for lang in sublist]

# Extract all languages wanted
want_to_work = df['LanguageWantToWorkWith'].dropna().str.split(';')
want_to_work = [lang for sublist in want_to_work for lang in sublist]

# Count frequency
have_worked_counts = pd.Series(have_worked).value_counts()
want_to_work_counts = pd.Series(want_to_work).value_counts()

# Combine into one DataFrame
language_df = pd.DataFrame({
    'WorkedWith': have_worked_counts,
    'WantToWorkWith': want_to_work_counts
}).fillna(0)

# Sort by 'WorkedWith' frequency
language_df = language_df.sort_values(by='WorkedWith', ascending=False)
language_df.head(10)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Top 10 languages by WorkedWith
top_languages = language_df.head(10)
x = np.arange(len(top_languages))

plt.figure(figsize=(12,6))
plt.bar(x - 0.2, top_languages['WorkedWith'], width=0.4, label='Have Worked With')
plt.bar(x + 0.2, top_languages['WantToWorkWith'], width=0.4, label='Want To Work With')
plt.xticks(x, top_languages.index, rotation=45)
plt.ylabel("Number of Respondents")
plt.title("Comparison of Programming Languages: Worked With vs Want To Work With")
plt.legend()
plt.show()
```

```python
!pip install matplotlib-venn


from matplotlib_venn import venn2

# Convert to sets
set_have = set(have_worked)
set_want = set(want_to_work)

plt.figure(figsize=(6,6))
venn2([set_have, set_want], set_labels=('Have Worked With', 'Want To Work With'))
plt.title("Overlap of Programming Languages")
plt.show()
```

### Step 7: Analyze Remote Work Trends


- Visualize the distribution of RemoteWork by region using a grouped bar chart or heatmap.

```python
import seaborn as sns

plt.figure(figsize=(12,8))
sns.heatmap(remote_country_counts.head(10), annot=True, fmt="d", cmap="YlGnBu")
plt.title("Remote Work Distribution by Country (Top 10 Countries)")
plt.xlabel("Remote Work")
plt.ylabel("Country")
plt.show()
```

### Step 8: Correlation between Job Satisfaction and Experience


- Analyze the correlation between overall job satisfaction (`JobSat`) and `YearsCodePro`.
  
- Calculate the Pearson or Spearman correlation coefficient.


```python
# Define mapping for JobSat
jobsat_mapping = {
    "Very satisfied": 5,
    "Slightly satisfied": 4,
    "Neither satisfied nor dissatisfied": 3,
    "Slightly dissatisfied": 2,
    "Very dissatisfied": 1
}

# Map JobSat to numeric
df['JobSat_numeric'] = df['JobSat'].map(jobsat_mapping)
```

```python
# Mapping special cases
years_mapping = {
    "Less than 1 year": 0.5,
    "More than 50 years": 51
}

# Apply mapping and convert to float
df['YearsCodePro_numeric'] = df['YearsCodePro'].replace(years_mapping)
df['YearsCodePro_numeric'] = pd.to_numeric(df['YearsCodePro_numeric'], errors='coerce')
```

```python
df_corr = df[['JobSat_numeric', 'YearsCodePro_numeric']].dropna()
```

```python
pearson_corr = df_corr['JobSat_numeric'].corr(df_corr['YearsCodePro_numeric'])
print("Pearson correlation between JobSat and YearsCodePro:", pearson_corr)
```

```python
spearman_corr = df_corr['JobSat_numeric'].corr(df_corr['YearsCodePro_numeric'], method='spearman')
print("Spearman correlation between JobSat and YearsCodePro:", spearman_corr)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))
sns.regplot(x='YearsCodePro_numeric', y='JobSat_numeric', data=df_corr, scatter_kws={'alpha':0.5})
plt.xlabel("Years of Professional Coding Experience")
plt.ylabel("Job Satisfaction (Numeric)")
plt.title("Job Satisfaction vs. Professional Coding Experience")
plt.show()
```

### Step 9: Cross-tabulation Analysis (Employment vs. Education Level)


- Analyze the relationship between employment status (`Employment`) and education level (`EdLevel`).

- **Instruction**: Create a cross-tabulation using `pd.crosstab()` and visualize it with a stacked bar plot if possible.


```python
import pandas as pd

# Cross-tab of Employment vs EdLevel
employment_edu_ct = pd.crosstab(df['EdLevel'], df['Employment'])

# Display the first few rows
employment_edu_ct.head()
```
